```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('board2.csv')
df['Average'] = df[['Math', 'Electronics', 'GEAS', 'Communication']].mean(axis=1)
df
```

## A. VISAYAS COMMUNICATION DATAFRAME

```python
VisComm = df[(df["Hometown"] == "Visayas") & (df["Track"] == "Communication")][
    ["Name", "Gender", "Math", "Electronics", "Average"]].copy()

print("VisComm")
display(VisComm)
print(f"Number of rows: {len(VisComm)}")
```

## B. VISAYAS FEMALE DATAFRAME

```python
VisFemale = df[(df["Hometown"] == "Visayas") & (df["Gender"] == "Female")][
    ["Name", "Track", "GEAS", "Electronics", "Average"]].copy()

print("VisFemale")
display(VisFemale)
print("\nVisFemale whose Average is at least 60")
display(VisFemale[VisFemale["Average"] >= 60])
```

## C. CATEGORY-AVERAGE VISUALIZATION

```python
track_avg = df.groupby("Track")["Average"].mean().reset_index()
gender_avg = df.groupby("Gender")["Average"].mean().reset_index()
hometown_avg = df.groupby("Hometown")["Average"].mean().reset_index()

display(track_avg)
display(gender_avg)
display(hometown_avg)

plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.bar(track_avg["Track"], track_avg["Average"])
plt.title("Mean Average by Track")

plt.subplot(1, 3, 2)
plt.bar(gender_avg["Gender"], gender_avg["Average"])
plt.title("Mean Average by Gender")

plt.subplot(1, 3, 3)
plt.bar(hometown_avg["Hometown"], hometown_avg["Average"])
plt.title("Mean Average by Hometown")

plt.tight_layout()
plt.show()

print("The Track with the highest sample mean average is Communication.")
print("The Gender with the highest sample mean average is Male.")
print("The Hometown with the highest sample mean average is Luzon.")
```
